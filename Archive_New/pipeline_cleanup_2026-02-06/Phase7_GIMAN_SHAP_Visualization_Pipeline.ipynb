```python
# 💾 DATA PERSISTENCE & CHECKPOINT SYSTEM
"""
Comprehensive save/load system for all GIMAN phase data, models, and results
Enables safe notebook restart from any checkpoint
"""

import pickle
import joblib
from datetime import datetime

# Create checkpoint directory structure
CHECKPOINT_DIR = Path("phase7_checkpoints")
CHECKPOINT_DIR.mkdir(exist_ok=True)

# Define checkpoint subdirectories
CHECKPOINT_SUBDIRS = {
    'data': CHECKPOINT_DIR / "data",
    'models': CHECKPOINT_DIR / "models", 
    'results': CHECKPOINT_DIR / "results",
    'visualizations': CHECKPOINT_DIR / "visualizations",
    'metadata': CHECKPOINT_DIR / "metadata"
}

# Create all subdirectories
for subdir_name, subdir_path in CHECKPOINT_SUBDIRS.items():
    subdir_path.mkdir(exist_ok=True)
    print(f"📁 Created checkpoint directory: {subdir_path}")

class GIMANCheckpointManager:
    """Comprehensive checkpoint management for GIMAN Phase 7 pipeline"""
    
    def __init__(self, checkpoint_dir: Path = CHECKPOINT_DIR):
        self.checkpoint_dir = checkpoint_dir
        self.metadata_file = checkpoint_dir / "metadata" / "checkpoint_metadata.json"
        self.load_metadata()
    
    def load_metadata(self):
        """Load existing checkpoint metadata or create new"""
        if self.metadata_file.exists():
            with open(self.metadata_file, 'r') as f:
                self.metadata = json.load(f)
            print(f"📊 Loaded existing checkpoint metadata: {len(self.metadata)} checkpoints")
        else:
            self.metadata = {}
            print("📊 Creating new checkpoint metadata")
    
    def save_metadata(self):
        """Save checkpoint metadata"""
        with open(self.metadata_file, 'w') as f:
            json.dump(self.metadata, f, indent=2)
    
    def save_checkpoint(self, name: str, data: dict, description: str = ""):
        """Save a comprehensive checkpoint"""
        timestamp = datetime.now().isoformat()
        
        # Save different data types appropriately
        checkpoint_info = {
            'timestamp': timestamp,
            'description': description,
            'files': {}
        }
        
        for key, value in data.items():
            if isinstance(value, pd.DataFrame):
                # Save DataFrames as CSV and pickle
                csv_path = CHECKPOINT_SUBDIRS['data'] / f"{name}_{key}.csv"
                pkl_path = CHECKPOINT_SUBDIRS['data'] / f"{name}_{key}.pkl"
                value.to_csv(csv_path, index=False)
                value.to_pickle(pkl_path)
                checkpoint_info['files'][key] = {'csv': str(csv_path), 'pkl': str(pkl_path), 'type': 'DataFrame'}
                
            elif isinstance(value, (np.ndarray, list)):
                # Save arrays/lists as numpy files
                npy_path = CHECKPOINT_SUBDIRS['data'] / f"{name}_{key}.npy"
                np.save(npy_path, value)
                checkpoint_info['files'][key] = {'npy': str(npy_path), 'type': 'array'}
                
            elif hasattr(value, 'state_dict'):  # PyTorch models
                # Save PyTorch models
                model_path = CHECKPOINT_SUBDIRS['models'] / f"{name}_{key}.pth"
                torch.save(value.state_dict(), model_path)
                checkpoint_info['files'][key] = {'pth': str(model_path), 'type': 'pytorch_model'}
                
            elif isinstance(value, dict):
                # Save dictionaries as JSON
                json_path = CHECKPOINT_SUBDIRS['results'] / f"{name}_{key}.json"
                with open(json_path, 'w') as f:
                    json.dump(value, f, indent=2, default=str)
                checkpoint_info['files'][key] = {'json': str(json_path), 'type': 'dict'}
                
            else:
                # Save other objects with pickle
                pkl_path = CHECKPOINT_SUBDIRS['data'] / f"{name}_{key}.pkl"
                with open(pkl_path, 'wb') as f:
                    pickle.dump(value, f)
                checkpoint_info['files'][key] = {'pkl': str(pkl_path), 'type': 'object'}
        
        # Update metadata
        self.metadata[name] = checkpoint_info
        self.save_metadata()
        
        print(f"💾 Checkpoint '{name}' saved successfully")
        print(f"📅 Timestamp: {timestamp}")
        print(f"📝 Description: {description}")
        print(f"📊 Data items saved: {len(data)}")
    
    def load_checkpoint(self, name: str) -> dict:
        """Load a comprehensive checkpoint"""
        if name not in self.metadata:
            print(f"❌ Checkpoint '{name}' not found")
            return {}
        
        checkpoint_info = self.metadata[name]
        loaded_data = {}
        
        print(f"📂 Loading checkpoint '{name}'...")
        print(f"📅 Created: {checkpoint_info['timestamp']}")
        print(f"📝 Description: {checkpoint_info['description']}")
        
        for key, file_info in checkpoint_info['files'].items():
            try:
                if file_info['type'] == 'DataFrame':
                    loaded_data[key] = pd.read_pickle(file_info['pkl'])
                elif file_info['type'] == 'array':
                    loaded_data[key] = np.load(file_info['npy'], allow_pickle=True)
                elif file_info['type'] == 'pytorch_model':
                    # Note: Model architecture needs to be recreated before loading state_dict
                    print(f"⚠️  PyTorch model '{key}' - state_dict path: {file_info['pth']}")
                    loaded_data[f"{key}_state_dict_path"] = file_info['pth']
                elif file_info['type'] == 'dict':
                    with open(file_info['json'], 'r') as f:
                        loaded_data[key] = json.load(f)
                elif file_info['type'] == 'object':
                    with open(file_info['pkl'], 'rb') as f:
                        loaded_data[key] = pickle.load(f)
                
                print(f"✅ Loaded: {key} ({file_info['type']})")
                
            except Exception as e:
                print(f"❌ Failed to load {key}: {e}")
        
        return loaded_data
    
    def list_checkpoints(self):
        """List all available checkpoints"""
        print(f"📋 AVAILABLE CHECKPOINTS ({len(self.metadata)}):")
        print("=" * 60)
        
        for name, info in self.metadata.items():
            print(f"📍 {name}")
            print(f"   📅 Created: {info['timestamp']}")
            print(f"   📝 Description: {info['description']}")
            print(f"   📊 Files: {len(info['files'])}")
            print()
    
    def cleanup_old_checkpoints(self, keep_last_n: int = 5):
        """Clean up old checkpoints, keeping only the most recent ones"""
        if len(self.metadata) <= keep_last_n:
            print(f"📊 Only {len(self.metadata)} checkpoints exist, no cleanup needed")
            return
        
        # Sort by timestamp and keep most recent
        sorted_checkpoints = sorted(
            self.metadata.items(), 
            key=lambda x: x[1]['timestamp'], 
            reverse=True
        )
        
        checkpoints_to_remove = sorted_checkpoints[keep_last_n:]
        
        for name, info in checkpoints_to_remove:
            # Remove files
            for file_info in info['files'].values():
                for file_path in file_info.values():
                    if isinstance(file_path, str) and Path(file_path).exists():
                        Path(file_path).unlink()
            
            # Remove from metadata
            del self.metadata[name]
            print(f"🗑️  Removed old checkpoint: {name}")
        
        self.save_metadata()
        print(f"✅ Cleanup complete. Kept {keep_last_n} most recent checkpoints")

# Initialize checkpoint manager
checkpoint_manager = GIMANCheckpointManager()

# Save initial timeline data as first checkpoint
initial_data = {
    'giman_timeline': GIMAN_TIMELINE,
    'phase6_available': PHASE6_INTERPRETABILITY_AVAILABLE,
    'phase6_outputs_available': PHASE6_OUTPUTS_AVAILABLE,
    'base_path': str(base_path)
}

checkpoint_manager.save_checkpoint(
    'initial_setup',
    initial_data,
    'Initial GIMAN timeline and phase availability data'
)

print("\n🎯 CHECKPOINT SYSTEM READY:")
print("=" * 50)
print("💾 All data will be automatically saved at key stages")
print("🔄 Notebook can be safely restarted from any checkpoint")
print("📊 Use checkpoint_manager.list_checkpoints() to see available saves")
print("📂 Use checkpoint_manager.load_checkpoint('name') to restore data")
```
